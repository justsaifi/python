```python
#read the file
with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
#write the file (overwritting)
with open('example.txt','w') as file:
    file.write('Hi\n')
    file.write('My name is Syed Saif Ali Shah\n')
```

```python
#append the file 
with open('example.txt','a') as file:
    file.write('i am in 7th semester')
```

```python
lines=['\nfirst line\n','second line\n','third line\n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
#copy from source to destination 
with open('example.txt','r') as file:
    content=file.read()
with open('destination.txt','w+') as file:
    file.write(content)
```

```python
### writing and then reading the file
with open('example.txt','a+') as file:
    file.writelines('my CGPA is 3.79')
    file.seek(0)
    content=file.read()

    print(content)
```
