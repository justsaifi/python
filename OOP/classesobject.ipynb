```python
#assignment 6
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
class Employee(Person):
    def __init__(self, name, age,employee_id):
        super().__init__(name, age)
        self.employee_id=employee_id
    def __str__(self):
        return f"employee name: {self.name} , employee age : {self.age} , Employee_id : {self.employee_id}"
    
person1=Employee('Abbas',21,201)
print(person1)
```

```python
#assignment 8
class Counter:
    count = 0

    def __init__(self):
        Counter.count += 1

    @classmethod
    def get_count(cls):
        return cls.count
c1=Counter()
c2=Counter()
print(c2.get_count())
```

```python
from abc import ABC,abstractmethod
class shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Circle(shape):
    def __init__(self,radius):
        self.radius=radius   
    def area(self):
        return math.pi * self.radius * self.radius
class Square(shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2
circle1=Circle(4)
print(circle1.area())
square1=Square(4)
print(square1.area())

```
