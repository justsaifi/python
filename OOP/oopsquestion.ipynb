```python
#assignment 1
import math

class shape:
    def area(self):
        pass
class circle(shape):
    def __init__(self,radius):
        self.radius=radius   
    def area(self):
        return 3.14 *self.radius**2
class square(shape):
    def __init__(self,side):
        self.side=side
    def area(self):
        return self.side*2
shapes=[circle(5),square(4)]
for shape in shapes:
    print(shape.area())
```

```python
#assignemnt 3
from abc import ABC,abstractmethod
class vehicle(ABC):
    @abstractmethod
    def engine(self):
        pass
class car(vehicle):
    def engine(self):
        print("car engine started")
class bike(vehicle):
    def engine(self):
        print('bike engine started')
c1=car()
c1.engine()
b1=bike()
b1.engine()

```

```python
#assignemnt 4
from abc import ABC,abstractmethod
class vehicle(ABC):
    @abstractmethod
    def engine(self):
        pass
    def fuel_type(self):
        return "generic fuel"
class car(vehicle):
    def engine(self):
        print("car engine started")
    def fuel_type(self):
            return "petrol"
class bike(vehicle):
    def engine(self):
        print('bike engine started')
    def fuel_type(self):
            return "deisel"
c1=car()
print(c1.fuel_type())
b1=bike()
print(b1.fuel_type())
```
